# ML-04 — Data Contract

## Dataset
The project uses the bundled anonymized release `data/raw/content_refresh_anonymized.csv`: 30,000 rows and 44 columns. The analysis unit is one anonymized content item.

## Inclusion / exclusion
Included rows must have positive 90-day impressions and content age ≥90 days; duplicates by `content_id` are removed. The starter preparation retained all 30,000 rows after these checks.

## Target and leakage
Target: `trend_direction == down`. Exclude `trend_direction`, `trend_pct`, `content_id`, and `client_id` from predictive features. IDs are used only for grouping/traceability. Do not use post-label information.

## Model feature groups
Numeric: search demand, competition, content size, historical volume, age/freshness, CTR, position and engagement signals.
Categorical: competition level, content type, intent, age/freshness tiers, word-count tier, impression tier and position tier.

```
from pathlib import Path
import pandas as pd
import numpy as np
ROOT = Path.cwd()
while ROOT.name and not (ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
DATA = ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv'
df = pd.read_csv(DATA)
print('Rows:', len(df), '| Columns:', len(df.columns))
required=['content_id','client_id','trend_direction','trend_pct','impressions_90d','content_age_days']
print('Missing required:', [c for c in required if c not in df.columns])
print('Duplicate content_id:', int(df.content_id.duplicated().sum()))
print('Positive impressions:', int((df.impressions_90d>0).sum()))
print('Age >= 90:', int((df.content_age_days>=90).sum()))
```
