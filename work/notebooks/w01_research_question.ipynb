# ML-02 — Research Question and Provisional Lane

## 1. My lane
**Refresh / Content Opportunity Scoring.** The project prioritizes existing pages for human review using observed performance, freshness, visibility and engagement signals. The goal is decision support, not automatic publishing.

## 2. Research question
**Which existing pages should be reviewed first for refresh, based on measurable signals associated with observed decline?**

The content/SEO team acts on the ranked queue. A wrong call can waste editorial time or cause a genuinely valuable page to be overlooked.

## 3. Quick look at the data

```
from pathlib import Path
import pandas as pd
import numpy as np
ROOT = Path.cwd()
while ROOT.name and not (ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
DATA = ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv'
df = pd.read_csv(DATA)
print('Rows:', len(df), '| Columns:', len(df.columns))
print('Declining-label rate:', round(df.trend_direction.eq('down').mean(),3))
print('Median impressions_90d:', int(df.impressions_90d.median()))
print('Median days_since_last_update:', int(df.days_since_last_update.median()))
```

## 4. Careful claims
This work can report **observed, measured, directional associations** and provide a ranked review queue. It cannot prove that refreshing a page causes ranking/traffic gains and it does not predict Google's algorithm.
