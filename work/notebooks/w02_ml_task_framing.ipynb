# ML-03 — Frame the Refresh Lane as an ML Task

## 1. ML task type
**Binary classification used for scoring/ranking.** The model estimates the probability that a page belongs to the observed declining class; those probabilities are then used to rank review candidates.

## 2. Target
`is_declining_label = (trend_direction == 'down')`. This is a defined observed label in the released dataset. `trend_direction` and `trend_pct` are excluded from model features because they directly define or encode the target.

## 3. Success metric
**Precision@50** is the primary operational metric because the team has limited review capacity and wants the top of the queue to contain a high share of declining pages. ROC-AUC and average precision are secondary discrimination metrics. The positive-class base rate is always reported.

## 5. Why ML can beat a fixed rule
Several signals interact: visibility, position, age, content depth, CTR and engagement. A model can combine them into a probability score while the transparent rule remains the baseline.
