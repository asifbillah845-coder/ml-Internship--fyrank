# ML-09 — Validation and Research Claim Audit

## 1. Methodology questions from the research context
The strongest claims in SEO research require careful label definitions and validation that matches the decision. Here, the label is an observed decline class and the evaluation is a grouped client holdout; therefore claims stay within this snapshot.

## 2. Honest split result
The reference evaluation used 27,675 training rows and 2,325 test rows under a client-holdout split. Random Forest Precision@50 was 0.740 versus 0.240 for the baseline on that same test set.

```
import json
from pathlib import Path
ROOT=Path.cwd()
while ROOT.name and not (ROOT/'outputs'/'model_results.json').exists() and ROOT!=ROOT.parent: ROOT=ROOT.parent
r=json.loads((ROOT/'outputs'/'model_results.json').read_text())
print('Split:',r['split_strategy'])
print('Train:',r['train_rows'],'Test:',r['test_rows'])
print('Positive base rate:',round(r['target_positive_rate'],3))
print('RF P@50:',r['models']['random_forest']['precision_at_50'])
print('Baseline P@50:',r['baseline']['baseline_precision_at_50'])
```

## 3. Leakage audit
The final model feature list excludes `trend_direction`, `trend_pct`, `content_id`, and `client_id`. This prevents direct target leakage and identifier memorization.

## 4. Claim rewrite
**Safe claim:** In the bundled anonymized snapshot, the Random Forest ranked a higher share of observed-declining pages in its top 50 than the transparent baseline under a client-holdout evaluation. This supports directional decision support for review prioritization; it does not establish that refreshing a page will cause a performance improvement.

## Limitation
No true future-window label is available in this snapshot, so this is not a production-grade temporal forecast validation.
