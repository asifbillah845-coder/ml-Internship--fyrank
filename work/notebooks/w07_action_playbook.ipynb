# ML-10 — Content Action Playbook

## 1. Ranked actions + reason codes
The final queue combines model probability with baseline/review signals. Actions are human-review recommendations: refresh, refresh-and-review-CTR, refresh-and-review-engagement, expand-and-refresh, or monitor.

```
import pandas as pd
from pathlib import Path
ROOT=Path.cwd()
while ROOT.name and not (ROOT/'outputs'/'refresh_queue.csv').exists() and ROOT!=ROOT.parent: ROOT=ROOT.parent
q=pd.read_csv(ROOT/'outputs'/'refresh_queue.csv')
print(q[['final_rank','final_refresh_score','confidence','suggested_action','final_reason_codes']].head(20).to_string(index=False))
```

## 2. Intended use
Use the queue to decide which pages deserve scarce editorial/SEO review capacity. A human should verify the page context, search intent, current business priority and whether a refresh is actually appropriate.

## 3. Human review / no-go list
Never automatically publish, delete, merge, redirect, or rewrite content from the score alone. Never expose client-identifying information. Do not interpret a high score as proof of a Google ranking cause.

## 4. Monitoring / retrain triggers
Re-run when the data release changes materially, the decline base rate shifts, feature distributions drift, or editorial outcomes show degraded Precision@50. Prefer a future-window evaluation when longitudinal labels become available.

## 5. Export
The generated `outputs/refresh_queue.csv` is the paper-facing ranked artifact. Public copies should retain only anonymized/public-safe fields.
