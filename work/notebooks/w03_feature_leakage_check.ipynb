# ML-05 — Feature Leakage Check

## Leakage rules
A feature is rejected if it directly defines the label, is a pseudonymous identifier, or would only be known after the outcome being predicted.

```
from pathlib import Path
import pandas as pd
import numpy as np
ROOT = Path.cwd()
while ROOT.name and not (ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
DATA = ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv'
df = pd.read_csv(DATA)
print('Rows:', len(df), '| Columns:', len(df.columns))
forbidden=['trend_direction','trend_pct','content_id','client_id']
print('Forbidden columns present:', [c for c in forbidden if c in df.columns])
print('Target:', df.trend_direction.eq('down').mean())
```

## Final decision
`trend_direction` and `trend_pct` are not in the model feature list. `content_id` and `client_id` are identifiers only. The remaining feature set is based on pre-existing page/performance descriptors in the released snapshot.

## Important limitation
Because this release is a snapshot rather than a longitudinal event table, the evaluation is a grouped client holdout rather than a future-time forecast.
